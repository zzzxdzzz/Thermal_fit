## STEP 1. Preprocessing of KRC Model

The temperature models are provided by T.N. Titus using KRC.  The models are in a multidimentional array, with 48 (Local Time) x 19 (Latitudes) x 100 (depths) x 7 (thermal inertia) x 3 (emissivity).  Parameters listed below, based on Titus's email on 2023-09-11.
* Local time starts at midnight in increments of 0.5 hours.
* Latitudes start at -90 and proceed to +90 in 10 degree increments.
* Thermal Inertia starts at 10, 20, 40, 80, 160, 320, 640 tiu.
* Emissivity 0.5, 0.7, 0.9.
* Depths are in meters of a skin depth specified in FITS extensions.

The mapping from temperature model to Didymos's and Dimorphos's simulated images using triangulate plate shape model is performed with IDL function 'tpm_sph2plt.pro'.  This function requires temperature model to be in a 3D array of (N, D, M) in N local solar times starting from local noon, D depths, and M latitudes.  Therefore, the original model needs to be processed to format to comply with this format.

In this step, the raw temperature models are processed and broken into individual temperature files that each corresponding to one combination of thermal inertia and emissivity.  Each temperature profile will have the format complying with the requirement by 'tpm_sph2plt.pro'.

The output are saved in file names `'KRC/[didy|dimo]/tref_gamma{:d}_emis{:.1f}.fits'.format(ti, emis)`

```python
# preprocess temperature model
# reshapes the 5D KRC temperature array into 3D slices per TI/emissivity combo and saves them in a format
# 5D KRC to 3D inputs for IDL 5D：Time × Lat × Depth × TI × Emis


from astropy.io import fits
from ALMACeres import tiu
u.add_enabled_units(tiu)

#obj = 'Ganymed'
obj = 'Ganymed'


if obj == 'Ganymed':
    infile = 'KRC/Ganymed_krc_temps.fits'
    outdir = 'KRC/Ganymed/'
    period_rot = 10.31284 * u.hour



# parameters from Titus:
period_orb = 4.35 * u.year    # orbital period   (from Tim:period_orb = 1589.339 (Julian Days))
rho = 11190 * u.Unit('kg/m3')    # density (from Tim/: Density  11190 kg/m3)
c = 750 * u.Unit('J/(K kg)')   # specific heat ( From Tim: Specific Heat 750)

# thermal skin depth
askin = np.sqrt(period_orb / np.pi) * tiu / rho * c

with fits.open(infile) as f_:
    temp = f_[0].data * u.K
    lst = f_[1].data * u.hour  #Local Solar Time
    lat = np.squeeze(f_[2].data) * u.deg 
    ti = np.squeeze(f_[3].data) * tiu # Thermal Inertia
    emis = np.squeeze(f_[4].data)   #emissivity
    dep = np.squeeze(f_[5].data) * u.m  #depth 

# make the format in compliance with the requirement by IDL routine 'tpm_sph2plt.pro'
temp = np.moveaxis(temp, 0, 3)  # move time axis to the first 
s = np.where(lst == 12 * u.hour)[0] # change starttime to noon for IDL
lst = np.roll(lst, -s)
if lst[0] != 12 * u.hour:
    raise ValueError('something went wrong with shifting arrays')
temp = np.roll(temp, -s, axis=-1)

for i, e in enumerate(emis):
    for j, t in enumerate(ti):
        # slice the temperature model
        tref = temp[i, j]

        # save reference temperature model
        hdu0 = fits.PrimaryHDU(tref.to_value('K'))
        hdu0.header['bunit'] = 'K'
        hdu0.header['ti'] = t.to_value('tiu'), 'thermal inertia [tiu]'
        hdu0.header['emiss'] = e, 'emissivity'
        hdu0.header['rho'] = rho.to_value('kg/m3'), 'density [kg/m**3]'
        hdu0.header['c'] = c.to_value('J/(K kg)'), 'specific heat [J/kg]'
        hdu0.header['p_orb'] = period_orb.to_value('year'), 'orbital period [year]'
        hdu0.header['p_rot'] = period_rot.to_value('hour'), 'rotational period [hour]'

        # t ** 2 ???? 
        hdu0.header['a_skin'] = (np.sqrt(period_orb / np.pi) * t / (rho * c)).to_value('m'), 'annual thermal skin depth [m]'
        hdu0.header['d_skin'] = (np.sqrt(period_rot / np.pi) * t / (rho * c)).to_value('m'), 'diurnal thermal skin depth [m]'
        hdu1 = fits.ImageHDU(lst.to_value('hour'), name='lst')
        hdu1.header['bunit'] = 'hour'
        hdu2 = fits.ImageHDU(dep[:, j].to_value('m'), name='depth') # seperated depth value
        hdu2.header['bunit'] = 'm'
        hdu3 = fits.ImageHDU(lat.to_value('deg'), name='lat')
        hdu3.header['bunit'] = 'deg'
        outfile = outdir + 'tref_gamma{:.0f}_emis{:.2f}.fits'.format(t.value, e)
        fits.HDUList([hdu0, hdu1, hdu2, hdu3]).writeto(outfile, overwrite=True)
```

## STEP 3 radiative transfer model 
1. explores a grid of dielectric properties (index of refraction and loss tangent) 
2. combine 100 depths

```python
# apply radiative transfer model to simulated images

# -- This script applies a radiative transfer (RT) model to a set of simulated thermal maps for Ganymed, 
# -- converting subsurface temperature profiles into surface emission fluxes (in Jy/arcsec²) at a given frequency. 
# -- It explores a grid of dielectric properties (index of refraction and loss tangent) and saves the results in new FITS files.

# use script 'rt_model.py' to do manual parallel computation.
#
#Notes： Ganymed at X band
#Notes：All fluxes are in Jy. The fits simultaneously sample frequencies from 7.976 GHz to 12.024 GHz. All times are UT
#We have about 1.5 hours of coverage at 75 GHz and 90 GHz on Sept 19, and about 20 minutes of coverage at 130 GHz and 145 GHz on Sept. 18
#Ka band gaussian fits for both dates, Ku gaussian fits for Oct 23, and an archival list of point source fits for Ku band on Oct 24 (ideally I’d like to fit gaussians). 

from glob import glob
from astropy.modeling.physical_models import BlackBody
from astropy.io import fits
from ALMACeres import Layer, Surface
import time

# observation frequency
freq = 10 * u.GHz   #---Change this accordingly---- all four add to this

wave = freq.to_value('m', u.spectral())

# index of refraction
nn = np.linspace(1.001, 2.5, 11)
#1-3 range


# loss tangent
loss = np.logspace(-4, 0, 9)

# the folder of the temp_maps
root = '/Users/zouxd/Documents/2-Projects/2024-YORPD/Ganymed/temp_maps2'
# get all the sub folders
subdirs = [os.path.join(root, name) for name in os.listdir(root)
           if os.path.isdir(os.path.join(root, name))]
# test print
#for d in subdirs:
#    print(d)


# fits keys to propagate
keys = ['ti', 'emiss', 'rho', 'c', 'p_orb', 'p_rot', 'a_skin', 'd_skin', 'long']


# loop through all subdirectories, corresponding to different TIs
for d in subdirs:
    print('1. d = ', d)
    #dd = '/'.join(d.split('/')[0:9]
    print('2. directory {}: starts on '.format('/'.join(d.split('/')[0:9])), end='')
    print(time.strftime("%b-%d %H:%M:%S", time.localtime()))
    
    t00 = time.time()
    
    # collect image files
    imfiles = glob(d + '/tempmap_???.fits')  # for lightcurve simulation

    # imfiles = glob(d + '/tempmap_simu??.fits')  # for ALMA images
    
    # loop through all images
    for f in imfiles:
        print('    simu image {}: '.format(os.path.basename(f)), end='')
        t0 = time.time()
        
        # load model
        with fits.open(f) as f_:
            intensity = BlackBody(f_[0].data * u.K)(freq).to_value('Jy/arcsec2')
            zz = f_[1].data
            emi = f_[2].data
            hdr = f_[0].header
        
        # loop through dielectric parameter space
        ims = np.zeros(nn.shape + loss.shape + intensity.shape[1:])
        for i in range(intensity.shape[-2]):
            for j in range(intensity.shape[-1]):
                if intensity[0, i, j] <= 0:
                    continue
                layer = Layer(1.5, 1e-2, profile=[zz, intensity[:, i, j]])
                surface = Surface(layer)
                for k, n in enumerate(nn):
                    for t, l in enumerate(loss):
                        surface.layers[0].n = n
                        surface.layers[0].loss_tangent = l
                        ims[k, t, i, j] = surface.emission(emi[i, j], wave)

        # save simulated images
        tmp = f.split('_')
        outfile = '_'.join(tmp[:-1] + ['rt'] + [tmp[-1]])
        hdu = fits.PrimaryHDU(ims.astype('float32'))
        hdu.header['bunit'] = 'Jy/arcsec2'
        for k in keys:
            hdu.header[k] = hdr[k], hdr.comments[k]
        hdu1 = fits.ImageHDU(nn.astype('float32'), name='refidx')
        hdu2 = fits.ImageHDU(loss.astype('float32'), name='loss')
        fits.HDUList([hdu, hdu1, hdu2]).writeto(outfile, overwrite=True)
        
        print('file processing time {:.2f} min'.format((time.time() - t0) / 60))
    print('directory processing ends on ', end='')
    print(time.strftime("%b-%d %H:%M:%S", time.localtime()), end=', ')
    print('total processing time {:.2f} min'.format((time.time() - t00) / 60))
    
```

```python
# For Ganymed integrated model data (fixed/robust)
import os
import re
import numpy as np
from glob import glob
from astropy.io import fits
import astropy.units as u

# --------------------
# User-editable inputs
# --------------------
folder_path = '/Users/zouxd/Documents/GitHub/Thermal_fit/Ganymed/temp_maps'
filename_glob = '*_rt_???.fits'       # pattern inside each folder  ------------------- 90??????????
outfile = 'Ganymed_integrated_model_90.fits'

# Observation / pixel geometry
freq = 343.5 * u.GHz                     # observation frequency
pxlscl = 0.1 * u.mas                     # pixel scale (assume square pixels)

# If you want to enforce a specific emissivity order in output, list it here;
# otherwise it will be inferred from folder names:
preferred_emis_order = [0.75, 0.85, 0.95]

# ------------------------------------------------
# Helpers: parse floats after labels in foldername
# ------------------------------------------------
def parse_float_after(label: str, text: str):
    # e.g., label='gamma', text='...gamma250_emis0.85...'
    m = re.search(rf'{re.escape(label)}\s*([-+]?\d*\.?\d+(?:[eE][-+]?\d+)?)', text)
    return float(m.group(1)) if m else None

# -----------------------
# Collect & parse folders
# -----------------------
if not os.path.isdir(folder_path):
    raise FileNotFoundError(f"Folder not found: {folder_path}")

all_dirs = [
    os.path.join(folder_path, name)
    for name in os.listdir(folder_path)
    if os.path.isdir(os.path.join(folder_path, name))
]

if not all_dirs:
    raise RuntimeError(f"No subdirectories found under {folder_path}")

records = []
for d in all_dirs:
    base = os.path.basename(d)
    g = parse_float_after('gamma', base)
    e = parse_float_after('emis',  base)
    if g is None:
        raise ValueError(f"Cannot parse 'gamma' from folder name: {base}")
    if e is None:
        # if emissivity is not in folder names and you know it elsewhere,
        # you could set a default; but better to be explicit:
        raise ValueError(f"Cannot parse 'emis' (emissivity) from folder name: {base}")
    records.append((g, e, d))

# Determine unique sets and sort
unique_gamma = sorted({g for g, e, d in records})
unique_emis  = sorted({e for g, e, d in records})

# If you prefer a particular emissivity order, reorder accordingly (only if all present)
if set(np.round(unique_emis, 5)) == set(np.round(preferred_emis_order, 5)):
    unique_emis = preferred_emis_order

# Map to index for stable sort (TI major, then emissivity)
g_index = {g: i for i, g in enumerate(unique_gamma)}
e_index = {e: i for i, e in enumerate(unique_emis)}

# Sort folders deterministically by (TI, emis)
records.sort(key=lambda t: (g_index[t[0]], e_index[t[1]]))

nti   = len(unique_gamma)
nemis = len(unique_emis)

# ----------------------------
# Loop through folders & files
# ----------------------------
px_omega = (pxlscl.to(u.rad))**2            # pixel solid angle in rad^2
px_omega_sr = px_omega.to(u.sr)             # explicitly as steradian

flux_all  = []   # will become shape: (nti*nemis, nlon, nref, nloss)
tb_all    = []
A_all     = []   # projected area per image (sr), shape: (nti*nemis, nlon)
gamma_out = []   # one TI per (TI,emis) row
emis_out  = []   # one emissivity per (TI,emis) row
lon_list  = None
nref = nloss = None

for g, e, d in records:
    # FITS files for this (TI,emis)
    simufiles = sorted(glob(os.path.join(d, filename_glob)))
    if not simufiles:
        raise RuntimeError(f"No FITS files matched in {d} with pattern {filename_glob}")

    # If longitude is encoded in the filename, you can parse & sort by it here.
    # Otherwise, sorted() by name is used as a proxy for lon order.
    # Build lon indices from file order (0..nlon-1) or parse actual values if available.
    nlon_local = len(simufiles)

    flux_d = []
    tb_d   = []
    A_d    = []

    for fpath in simufiles:
        with fits.open(fpath) as hdul:
            im = hdul[0].data
            bunit = u.Unit(hdul[0].header.get('bunit', 'Jy / sr'))
            im = im * bunit  # e.g., surface brightness [Jy/sr] or similar

            # Read reference arrays (assume present and consistent)
            if 'REFIDX' in hdul:
                nn_local = np.array(hdul['REFIDX'].data)
            elif 'refidx' in hdul:
                nn_local = np.array(hdul['refidx'].data)
            else:
                raise KeyError(f"'REFIDX' extension not found in {fpath}")

            if 'LOSS' in hdul:
                loss_local = np.array(hdul['LOSS'].data)
            elif 'loss' in hdul:
                loss_local = np.array(hdul['loss'].data)
            else:
                raise KeyError(f"'LOSS' extension not found in {fpath}")

        # Confirm shape convention: (nref, nloss, ny, nx)
        if im.ndim != 4:
            raise ValueError(f"Expected image with 4 dims (nref,nloss,ny,nx). Got {im.shape} in {fpath}")

        if nref is None:
            nref = im.shape[0]
            nloss = im.shape[1]
        else:
            if (nref, nloss) != (im.shape[0], im.shape[1]):
                raise ValueError(f"Inconsistent (nref,nloss): {im.shape[:2]} vs ({nref},{nloss}) in {fpath}")

        # Integrated flux density over pixels: sum(I_nu [Jy/sr]) * dOmega [sr] -> [Jy]
        # sum over x,y (last two axes)
        flux = (im.sum(axis=-1).sum(axis=-1) * px_omega_sr).to(u.Jy)  # shape (nref, nloss)

        # Projected area = number of 'on-disk' pixels * dOmega; detect 'on-disk' as >0
        on_disk = (im[0, 0] > 0)  # boolean mask from one channel
        n_pix = int(on_disk.astype(int).sum())
        area_sr = (n_pix * px_omega_sr).to(u.sr)

        # Average brightness temperature estimate:
        # Treat the integrated flux as if from a "beam" of area = projected area,
        # so "flux/beam" == average over the disk.
        # Using astropy equivalency: Jy/beam <-> K at freq, beam area.
        avg_flux_per_beam = flux                      # Jy per "beam" (whole-disk area)
        tb = avg_flux_per_beam.to(
            u.K,
            equivalencies=u.brightness_temperature(freq, area_sr)
        )  # shape (nref, nloss)

        flux_d.append(flux)      # (nref, nloss)
        tb_d.append(tb)          # (nref, nloss)
        A_d.append(area_sr)      # scalar (sr)

        # Build lon list from file order once
    if lon_list is None:
        lon_list = list(range(0, 360, 90))[:nlon_local]

    # Stack along longitude dimension -> (nlon, nref, nloss)
    flux_d = u.Quantity(flux_d)   # (nlon, nref, nloss)
    tb_d   = u.Quantity(tb_d)     # (nlon, nref, nloss)
    A_d    = u.Quantity(A_d)      # (nlon,) in sr

    flux_all.append(flux_d)
    tb_all.append(tb_d)
    A_all.append(A_d)
    gamma_out.append(g)
    emis_out.append(e)

# Convert lists to arrays
flux_all = u.Quantity(flux_all)   # (nti*nemis, nlon, nref, nloss)
tb_all   = u.Quantity(tb_all)     # same
A_all    = u.Quantity(A_all)      # (nti*nemis, nlon)
gamma_out = np.array(gamma_out)   # (nti*nemis,)
emis_out  = np.array(emis_out)    # (nti*nemis,)

# Reshape to (nti, nemis, ...)
# records were sorted by (TI,emis) so reshape is consistent
try:
    flux_all = flux_all.reshape(nti, nemis, flux_all.shape[1], nref, nloss)
    tb_all   = tb_all.reshape(nti, nemis, tb_all.shape[1], nref, nloss)
    A_all    = A_all.reshape(nti, nemis, A_all.shape[1])   # (nti, nemis, nlon)
    gamma_2d = gamma_out.reshape(nti, nemis)
    emis_2d  = emis_out.reshape(nti, nemis)
except Exception as e:
    raise RuntimeError(f"Reshape failed; check folder counts & parsing. Details: {e}")

# Sort by increasing TI (should already be, but do it defensively)
st = np.argsort(gamma_2d[:, 0])
flux_all = flux_all[st]
tb_all   = tb_all[st]
A_all    = A_all[st]
gamma_sorted = gamma_2d[st][:, 0]       # (nti,)
emis_sorted  = emis_2d[st][0, :]        # (nemis,)

# Cross-section (projected area) averaged over emissivity -> shape (nlon, nti)
A_reshaped = A_all.mean(axis=1).T       # (nlon, nti)

# ---------------
# Build FITS HDUs
# ---------------
hdu0 = fits.PrimaryHDU(tb_all.value)     # data: K
hdu0.header['BUNIT']   = 'K'
hdu0.header['AX1_NAME'] = 'loss'    # axis order: (ti, emis, lon, refidx, loss) in our arrays?
hdu0.header['AX2_NAME'] = 'refidx'  # NOTE: Our data array order is (ti, emis, lon, ref, loss)
hdu0.header['AX3_NAME'] = 'lon'
hdu0.header['AX4_NAME'] = 'emis'
hdu0.header['AX5_NAME'] = 'ti'
# To be extra clear, record the exact shape:
hdu0.header['SHAPE0'] = tb_all.shape[0]  # nti
hdu0.header['SHAPE1'] = tb_all.shape[1]  # nemis
hdu0.header['SHAPE2'] = tb_all.shape[2]  # nlon
hdu0.header['SHAPE3'] = tb_all.shape[3]  # nref
hdu0.header['SHAPE4'] = tb_all.shape[4]  # nloss

hdu1 = fits.ImageHDU(flux_all.to(u.Jy).value, name='FLUX')
hdu1.header['BUNIT'] = 'Jy'

# Use the last-read nn/loss; they should be consistent across files/folders
hdu2 = fits.ImageHDU(loss_local.astype(np.float64), name='LOSS')
hdu3 = fits.ImageHDU(nn_local.astype(np.float64),   name='REFIDX')

hdu4 = fits.ImageHDU(np.asarray(lon_list, dtype=np.float64), name='LON')
hdu4.header['BUNIT'] = 'deg'

hdu5 = fits.ImageHDU(np.asarray(emis_sorted, dtype=np.float64), name='EMIS')

hdu6 = fits.ImageHDU(np.asarray(gamma_sorted, dtype=np.float64), name='TI')
hdu6.header['BUNIT'] = 'tiu'

hdu7 = fits.ImageHDU(A_reshaped.to(u.sr).value, name='XSEC')
hdu7.header['BUNIT'] = 'sr'
hdu7.header['DESC']  = 'Projected area per (lon,ti) averaged over emissivity'

fits.HDUList([hdu0, hdu1, hdu2, hdu3, hdu4, hdu5, hdu6, hdu7]).writeto(outfile, overwrite=True)

print(f"Saved: {outfile}")
```

```python
#I need to figure out the 90 ----------------
import os, re
import numpy as np
from glob import glob
from astropy.io import fits
import astropy.units as u

# ----------------- Inputs -----------------
lon = list(range(0, 360, 90))[:2]  # [0, 90]
filename_glob = '*_rt_???.fits'
outfile = 'Ganymed_integrated_model_90.fits'

freq = 343.5 * u.GHz
pxlscl = 0.1 * u.mas  # pixel scale (square pixels assumed)

# Expected model grid (used for validation/reshape)
nemis_expected = 3
nti_expected = 7
nlon_expected = 2

folder_path = '/Users/zouxd/Documents/GitHub/Thermal_fit/Ganymed/temp_maps'

# ---------- helpers ----------
def parse_float_after(label: str, text: str):
    m = re.search(rf'{re.escape(label)}\s*([-+]?\d*\.?\d+(?:[eE][-+]?\d+)?)', text)
    return float(m.group(1)) if m else None

def get_ext(hdul, names):
    for nm in names:
        if nm in hdul:
            return hdul[nm].data
    raise KeyError(f"Missing extensions {names}")

# Pixel solid angle in sr
px_area_sr = (pxlscl.to(u.rad))**2
px_area_sr = px_area_sr.to(u.sr)

# ------------ collect folders and sort by (gamma, emis) ------------
if not os.path.isdir(folder_path):
    raise FileNotFoundError(folder_path)

simudir = [os.path.join(folder_path, name) for name in os.listdir(folder_path)
           if os.path.isdir(os.path.join(folder_path, name))]

if not simudir:
    raise RuntimeError(f"No subdirectories in {folder_path}")

records = []
for d in simudir:
    base = os.path.basename(d)
    g = parse_float_after('gamma', base)
    e = parse_float_after('emis', base)
    if g is None:
        raise ValueError(f"Cannot parse gamma from folder: {base}")
    if e is None:
        raise ValueError(f"Cannot parse emis from folder: {base}")
    records.append((g, e, d))

# Stable order: TI major, then emissivity
records.sort(key=lambda t: (t[0], t[1]))
gammas = sorted({g for g, e, d in records})
emises = sorted({e for g, e, d in records})

# --------- storage ---------
flux_all = []   # (nlon, nref, nloss) per (ti,emis)
tb_all   = []
A_all    = []   # (nlon,) projected area (sr) per (ti,emis)
gamma_all = []
emis_all  = []

nref = None
nloss = None
for g, e, d in records:
    print(f'Processing {os.path.basename(d)}')
    files = sorted(glob(os.path.join(d, filename_glob)))
    if len(files) != nlon_expected:
        raise ValueError(f"Expected {nlon_expected} files in {d}, found {len(files)}")

    flux_d = []
    tb_d = []
    A_d = []

    for fpath in files:
        with fits.open(fpath) as hdul:
            # primary image cube assumed (nref, nloss, ny, nx)
            im = hdul[0].data
            bunit = u.Unit(hdul[0].header.get('BUNIT', hdul[0].header.get('bunit', 'Jy/sr')))
            im = im * bunit

            nn = get_ext(hdul, ['REFIDX', 'refidx'])
            loss = get_ext(hdul, ['LOSS', 'loss'])

        if im.ndim != 4:
            raise ValueError(f"{fpath}: expected 4D (nref,nloss,ny,nx), got {im.shape}")

        # Check consistent (nref, nloss)
        if nref is None:
            nref, nloss = im.shape[0], im.shape[1]
        else:
            if (nref, nloss) != (im.shape[0], im.shape[1]):
                raise ValueError(f"Inconsistent (nref,nloss) in {fpath}: {im.shape[:2]} vs {(nref,nloss)}")

        # Integrated flux density: sum over y,x and multiply by pixel sr -> Jy
        flux = (im.sum(axis=-1).sum(axis=-1) * px_area_sr).to(u.Jy)   # shape (nref, nloss)

        # Projected area (sr): count "on-disk" pixels from a representative channel
        on_disk = (im[0, 0] > 0)
        area_sr = (on_disk.astype(int).sum() * px_area_sr).to(u.sr)

        # Disk-averaged brightness temp: treat whole disk as one "beam" of area = area_sr
        tb = flux.to(u.K, equivalencies=u.brightness_temperature(freq, area_sr))  # (nref, nloss)

        flux_d.append(flux)
        tb_d.append(tb)
        A_d.append(area_sr)

    flux_all.append(u.Quantity(flux_d))  # (nlon, nref, nloss)
    tb_all.append(u.Quantity(tb_d))      # (nlon, nref, nloss)
    A_all.append(u.Quantity(A_d))        # (nlon,)
    gamma_all.append(g)
    emis_all.append(e)

# Convert to arrays
flux_all = u.Quantity(flux_all)  # (nti*nemis, nlon, nref, nloss)
tb_all   = u.Quantity(tb_all)
A_all    = u.Quantity(A_all)     # (nti*nemis, nlon)
gamma_all = np.array(gamma_all)
emis_all  = np.array(emis_all)

# Validate counts and reshape to (nti, nemis, ...)
nti = len(gammas)
nemis = len(emises)
if nti != nti_expected or nemis != nemis_expected:
    raise RuntimeError(f"Found grid (ti={nti}, emis={nemis}) != expected ({nti_expected}, {nemis_expected}).")

try:
    flux_all = flux_all.reshape(nti, nemis, nlon_expected, nref, nloss)
    tb_all   = tb_all.reshape(nti, nemis, nlon_expected, nref, nloss)
    A_all    = A_all.reshape(nti, nemis, nlon_expected)
except Exception as exc:
    raise RuntimeError(f"Reshape failed; check folder grouping. {exc}")

# Ensure TI axis is increasing (it already is by our sort, but be defensive)
gamma_grid = gamma_all.reshape(nti, nemis)[:, 0]
st = np.argsort(gamma_grid)
flux_all = flux_all[st]
tb_all   = tb_all[st]
A_all    = A_all[st]
gamma_sorted = gamma_grid[st]
emis_sorted  = np.array(emises, dtype=float)

# XSEC: average over emissivity -> (nlon, nti)
A_reshaped = A_all.mean(axis=1).T  # (nlon, nti)

# ---------------- FITS ----------------
hdu0 = fits.PrimaryHDU(tb_all.to_value(u.K))
hdu0.header['BUNIT'] = 'K'
# Axis order recorded for clarity: (ti, emis, lon, refidx, loss)
hdu0.header['AX1_NAME'] = 'loss'
hdu0.header['AX2_NAME'] = 'refidx'
hdu0.header['AX3_NAME'] = 'lon'
hdu0.header['AX4_NAME'] = 'emis'
hdu0.header['AX5_NAME'] = 'ti'
hdu0.header['SHAPE0'] = tb_all.shape[0]
hdu0.header['SHAPE1'] = tb_all.shape[1]
hdu0.header['SHAPE2'] = tb_all.shape[2]
hdu0.header['SHAPE3'] = tb_all.shape[3]
hdu0.header['SHAPE4'] = tb_all.shape[4]

hdu1 = fits.ImageHDU(flux_all.to_value(u.Jy), name='FLUX')
hdu1.header['BUNIT'] = 'Jy'

hdu2 = fits.ImageHDU(loss.astype(float), name='LOSS')
hdu3 = fits.ImageHDU(nn.astype(float),   name='REFIDX')

hdu4 = fits.ImageHDU(np.asarray(lon, dtype=float), name='LON')
hdu4.header['BUNIT'] = 'deg'

hdu5 = fits.ImageHDU(emis_sorted.astype(float), name='EMIS')

hdu6 = fits.ImageHDU(gamma_sorted.astype(float), name='TI')
hdu6.header['BUNIT'] = 'tiu'

hdu7 = fits.ImageHDU(A_reshaped.to_value(u.sr), name='XSEC')
hdu7.header['BUNIT'] = 'sr'
hdu7.header['DESC']  = 'Projected area per (lon,ti) averaged over emissivity'

fits.HDUList([hdu0, hdu1, hdu2, hdu3, hdu4, hdu5, hdu6, hdu7]).writeto(outfile, overwrite=True)
print(f"Saved: {outfile}")
```

## Effective Emissivity

The effective emissivity is calculated here as the ratio of the measured flux to the modeled flux from the surface.

7/5/2023:

The concept of "effective emissivity" calculated here is confusing.  The traditional way of calculating the "mm-wavelength effective emissivity" is to use a spherical shape of the same effective size and a non-rotating body, i.e., STM, insolated and viewed at the same geometry to calculate the expected flux (e.g., [Redman et al. 1998](https://iopscience.iop.org/article/10.1086/300495)).  But the intention here is to study the rotational variations in thermal and dielectric properties after taking into account the effect of shape.  Therefore, we cannot use the traditional concept of effective emissivity here.

The go-around is to not invoke effective emissivity, but rather just use deviation from the best-fit model in the total flux.  The notations in the following cells are not changed.  But the interpretation should not be effective emissivity.

Below two cells are the original and obsolete now.

```python
from glob import glob
from astropy.modeling.models import BlackBody
import astropy.units as u
from astropy.io import ascii

obsfile = 'hunter_etal_table.ecsv'
geofile = 'data/juno_data_geom.csv'
simufile = glob('./temp_maps/tref_gamma40_emis0.8/tempmap_simu??.fits')

freq = 233 * u.GHz
pxlscl = 2 * u.mas  # pixel scale of simu images

# calculate model flux from surface temperature models
temp = u.Quantity([fits.open(f)[0].data[0] for f in simufile], u.K)
flux_model = (BlackBody(temp)(freq).to('Jy/arcsec2').sum(axis=1).sum(axis=1) * pxlscl**2).to('mJy')

# load observations
obs = ascii.read(obsfile)
geom = ascii.read(geofile)

# emissivity
eff_emis = (obs['flux'] / flux_model).decompose()
eff_emis_err = (obs['flux_err'] / flux_model).decompose()
```
